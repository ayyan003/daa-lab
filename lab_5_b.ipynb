```python
import random


# Partition function
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


# QuickSelect function
def quickselect(arr, low, high, k):
    if low <= high:
        pivot_index = partition(arr, low, high)

        if pivot_index == k:
            return arr[pivot_index]

        elif pivot_index > k:
            return quickselect(arr, low, pivot_index - 1, k)

        else:
            return quickselect(arr, pivot_index + 1, high, k)


# ---------------- Main Program ----------------

# Generate 1000 random integers
arr = [random.randint(1, 10000) for _ in range(1000)]

n = len(arr)

# k values
k1 = 1          # Minimum
k2 = n          # Maximum
k3 = n // 2     # Median

minimum = quickselect(arr.copy(), 0, n - 1, k1 - 1)
maximum = quickselect(arr.copy(), 0, n - 1, k2 - 1)
median = quickselect(arr.copy(), 0, n - 1, k3 - 1)

print("Array Size:", n)
print("Minimum Element (k=1):", minimum)
print("Median Element (k=n//2):", median)
print("Maximum Element (k=n):", maximum)
```
