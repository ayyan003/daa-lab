```python
import random

def find_min_secondmin_max_secondmax(arr):
    # Sort the array
    arr = sorted(arr)

    minimum = arr[0]
    second_minimum = arr[1]

    maximum = arr[-1]
    second_maximum = arr[-2]

    return minimum, second_minimum, maximum, second_maximum


# ---------------- Main Program ----------------

arr = [15, 8, 22, 3, 10, 27, 5, 19, 12]

print("Array:", arr)

minimum, second_minimum, maximum, second_maximum = find_min_secondmin_max_secondmax(arr)

print("\nResults:")
print("Minimum Element       :", minimum)
print("Second Minimum Element:", second_minimum)
print("Maximum Element       :", maximum)
print("Second Maximum Element:", second_maximum)

# Comparison Analysis
n = len(arr)

# Approximate comparisons
min_max_comparisons = (3 * n) // 2 - 2
extra_comparisons = 2

print("\nComparison Analysis")
print("----------------------------")
print("Elements               :", n)
print("Min-Max Comparisons    :", min_max_comparisons)
print("Additional Comparisons :", extra_comparisons)
print("Total Comparisons      :", min_max_comparisons + extra_comparisons)



```
