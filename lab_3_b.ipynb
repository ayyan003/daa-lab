```python
import sys

def prim_mst(graph): V = len(graph)

selected = [False] * V
key = [sys.maxsize] * V
parent = [-1] * V

# Start from vertex 0
key[0] = 0

for _ in range(V):
    # Find the minimum key vertex not yet included
    min_key = sys.maxsize
    u = -1

    for v in range(V):
        if not selected[v] and key[v] < min_key:
            min_key = key[v]
            u = v

    selected[u] = True

    # Update adjacent vertices
    for v in range(V):
        if graph[u][v] > 0 and not selected[v] and graph[u][v] < key[v]:
            key[v] = graph[u][v]
            parent[v] = u

print("Edge \tWeight")
total = 0
for i in range(1, V):
    print(f"{parent[i]} - {i}\t{graph[i][parent[i]]}")
    total += graph[i][parent[i]]

print("Total Weight =", total)



```
