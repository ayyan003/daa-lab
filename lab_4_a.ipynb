```python
import heapq

def dijkstra(graph, source):
    n = len(graph)

    dist = [float('inf')] * n
    prev = [None] * n
    dist[source] = 0

    pq = [(0, source)]  # (distance, vertex)

    while pq:
        d, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, w in graph[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, prev


def reconstruct_path(prev, source, target):
    path = []

    while target is not None:
        path.append(target)
        target = prev[target]

    path.reverse()

    if path[0] == source:
        return path
    return []


# 5-node city road network
graph = {
    0: [(1, 10), (2, 3)],   # Airport
    1: [(2, 1), (3, 2)],
    2: [(1, 4), (3, 8), (4, 2)],
    3: [(4, 7)],
    4: []                   # Central Station
}

source = 0      # Airport
target = 4      # Central Station

dist, prev = dijkstra(graph, source)

path = reconstruct_path(prev, source, target)

print("Fastest Route from Airport to Central Station")
print("--------------------------------------------")
print("Shortest Distance:", dist[target], "minutes")
print("Path:", " -> ".join(map(str, path)))
```
